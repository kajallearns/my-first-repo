```python
import random

num = random.randint(1, 10)
chances=3

while chances>0:
  try:
    guess = int(input("Enter your guess: "))
    if 1<=guess<=10:
      if num == guess:
          print("You guessed it!")
          break
      else:
          print("Wrong guess!")
          if guess>num:
            print("too high")
          else:
            print("too low")
          chances-=1
          print(f"you have {chances} attempts left" if chances>0 else "you ran out of all attempts")
    else:
      print("please enter a valid number between 1-10")

  except ValueError:
    print("please enter a valid number between 1-10")
```
